```python
# Plot 1 - Distribution of PRA Categories 
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("clean-nba.csv")

plt.figure(figsize=(8, 5))
df['PRA_Category'].value_counts().sort_index().plot(kind='bar')
plt.title("Distribution of PRA Categories")
plt.xlabel("PRA Category")
plt.ylabel("Number of Players")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Plot 2 - PRA Category Count by Age Group
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("clean-nba.csv")

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Age_Group', hue='PRA_Category', order=["Rookie", "Prime", "Veteran"])
plt.title("PRA Category Count by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Players")
plt.legend(title="PRA Category")
plt.tight_layout()
plt.show()
```
